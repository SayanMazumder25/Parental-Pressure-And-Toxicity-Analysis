## **2. Research Methodology and Study Design**
### 2.1 Research Design

This study adopts an exploratory, cross-sectional survey design to investigate students' perceptions of parental pressure, personal autonomy, and the overall parenting environment in the Indian educational context.

The study focuses on students' self-reported experiences of parental expectations, involvement in personal decisions, academic pressure, and related aspects of family interactions. It aims to identify patterns and associations between these experiences and students' reported perceptions of parental pressure and parenting environment.

The study combines descriptive statistical analysis, exploratory association analysis, and machine learning techniques. Descriptive analysis is used to summarise the characteristics of the respondents and their survey responses. Statistical methods are used to explore relationships between selected variables, while machine learning models are considered for predicting students' reported ratings from available demographic and survey-response data.

As the study is based on observational survey responses collected at a single point in time, its findings are interpreted as associations rather than evidence of causal relationships.

### 2.2 Study Population and Sampling

The intended population comprises students at different educational levels in India, including school students, undergraduate students, and postgraduate students.

Data were collected through an online questionnaire distributed to potential respondents. Participation was voluntary, and the resulting dataset constitutes a non-probability sample.

Consequently, the sample may not represent the broader population of Indian students. Differences in access to the survey, willingness to participate, and respondents' educational backgrounds may influence the composition of the dataset.

The final sample size and the distribution of respondents across demographic categories will be reported after the collected responses have been verified. Particular attention will be given to the number of observations available in each educational and demographic group when conducting comparative analyses.

### 2.3 Survey Instrument

Primary data were collected using a structured questionnaire designed to capture students' perceptions of parental expectations, pressure, personal autonomy, and parenting environment.

The questionnaire contains demographic questions, statements concerning students' reported experiences, and overall rating questions.

The demographic variables include age, gender, educational level, and residential background. The experience-related statements address aspects of parental behaviour and student experiences, including academic expectations, comparisons with others, influence over career choices, involvement in personal decisions, restrictions on independence, interference in social relationships, feelings of guilt when disagreeing with parents, dismissal of students' opinions, and fear of disappointing parents.

The experience-related statements use a five-point frequency scale:

| Response | Numerical value |
|---|---:|
| Never | 1 |
| Rarely | 2 |
| Sometimes | 3 |
| Often | 4 |
| Very Often | 5 |

The questionnaire also includes two overall ratings: perceived parental pressure and perceived parenting environment. These ratings use a ten-point scale, with the parenting-environment scale running from a very healthy or supportive environment to a very toxic or unsupportive environment.

The original questionnaire and its response labels will be retained as the reference for data coding and interpretation. Before analysis, the exported dataset will be checked to ensure that the recorded values correspond to the intended questions and response scales.

### 2.4 Variables and Analytical Framework

The analysis distinguishes between demographic characteristics, individual survey responses, and overall reported outcomes.

| Variable category | Variables | Analytical purpose |
|---|---|---|
| Demographic characteristics | Age, gender, educational level, residential background | Describe the sample and explore demographic associations |
| Experience-related responses | Individual parental-behaviour and student-experience statements | Identify reported experiences associated with parental pressure |
| Overall outcomes | Parental-pressure rating and parenting-environment rating | Examine the reported outcomes and explore their predictability |

The two overall ratings will be analysed separately because they represent related but distinct aspects of students' perceptions.

The parental-pressure rating reflects respondents' overall assessment of the pressure they experience. The parenting-environment rating reflects their overall assessment of the family environment according to the questionnaire's specified scale.

The individual experience-related responses will be examined to identify their associations with these outcomes. However, the analysis will distinguish statistical association from causal influence and will avoid interpreting any single response as a definitive measure of family functioning.

Before predictive modelling, the construction of both outcome variables will be verified. If either outcome is directly calculated from the individual survey items, those same items will not be used as predictors for that outcome without an appropriate methodological justification, because doing so could introduce target leakage.

### 2.5 Data Analysis Strategy

The analysis will proceed in several stages.

**Stage 1: Data inspection and preprocessing**

The exported survey data will be inspected for missing values, duplicate records, inconsistent categories, invalid responses, and data-type issues. The original dataset will be preserved, and preprocessing will be conducted on a separate working copy.

**Stage 2: Descriptive analysis**

Descriptive statistics and visualisations will be used to summarise the demographic composition of the sample and the distribution of responses to individual survey questions and overall ratings.

**Stage 3: Exploratory association analysis**

Suitable statistical techniques will be used to investigate relationships between selected survey responses, demographic characteristics, and overall ratings. Spearman's rank correlation may be used for appropriate ordinal or numeric variables. Group-comparison methods may be considered when the research question, independence of observations, and available group sizes justify their use.

**Stage 4: Comparative analysis**

The study will explore whether the reported ratings differ across selected demographic groups, including gender and educational level. Group sizes and rating distributions will be examined before formal statistical comparisons are undertaken.

**Stage 5: Machine learning**

If the data quality and sample size permit, regression models will be developed to explore whether demographic characteristics and eligible survey responses can predict the two overall ratings. Model performance will be compared against a simple baseline using appropriate regression metrics.

The modelling results will be interpreted as exploratory predictive findings rather than proof of causal relationships or evidence that the models will generalise to the wider student population.

### 2.6 Ethical Considerations

The study concerns potentially sensitive experiences involving family relationships, academic expectations, personal autonomy, and perceived parental pressure. Accordingly, the analysis will be conducted with attention to respondent privacy and responsible interpretation.

The dataset will be checked for directly identifying information before it is shared or incorporated into public research materials. Results will primarily be presented in aggregate form, and small demographic groups will be handled cautiously to reduce the risk of inadvertently identifying individual respondents.

The study will not use its findings to diagnose individual families or classify particular parents as toxic. Instead, it seeks to understand patterns in the experiences reported by participating students.

Any applicable institutional requirements concerning research ethics, informed consent, and participation by minors must be verified and followed before the findings are disseminated. No claim of formal ethical approval will be made unless such approval has actually been obtained.

### 2.7 Methodological Limitations

Several limitations are relevant to the interpretation of the study.

First, voluntary participation and non-probability sampling may introduce selection bias. The experiences of respondents may differ from those of students who did not participate.

Second, the study relies on self-reported responses. These responses reflect participants' perceptions and may be influenced by individual interpretations, recall, or willingness to disclose sensitive experiences.

Third, the cross-sectional design does not establish temporal ordering or causality. Observed associations cannot demonstrate that a particular parental behaviour caused a specific outcome.

Fourth, some demographic groups may contain substantially fewer respondents than others. Such imbalances can make comparisons less precise and may limit the reliability of predictive models.

Finally, machine learning performance estimated from a relatively small dataset may be unstable and may not generalise to new populations. Predictive performance will therefore be assessed cautiously, with attention to validation procedures, model limitations, and the possibility of overfitting.

### 2.8 Summary

This methodology establishes a framework for examining students' reported experiences of parental pressure and their perceptions of the parenting environment using survey data and computational analysis.

The study combines descriptive statistics, exploratory association analysis, demographic comparisons, and carefully evaluated machine learning models. Its principal objective is to identify interpretable patterns in the collected responses while maintaining appropriate caution regarding sampling limitations, uncertainty, and causal interpretation.

## **6 Research Question 1: Which Experiences are Associated with Parental Pressure and Toxicity?**
**6.1 Introduction and Methodology**

To answer the first research question, the analysis examines the relationships between nine survey items and two outcome variables: overall parental pressure (Pressure_Evaluation) and perceived parental toxicity (Toxicity_Relevance).

Spearman's rank correlation coefficient was used to assess the direction and strength of the associations between each survey item and each outcome. This method is suitable for examining relationships between ordinal survey ratings without requiring the relationships to be linear.

Two correlation visualizations are presented below. The first illustrates the associations between the nine survey items and overall parental pressure, while the second illustrates their associations with perceived parental toxicity. Comparing the two visualizations helps identify which reported experiences are more closely associated with each outcome and whether the patterns differ between them.

These correlations indicate associations rather than causal relationships. Therefore, the analysis identifies experiences that tend to accompany higher outcome ratings, rather than establishing the reasons why parental pressure or toxicity occurs.

```python
df1.columns = df1.columns.str.strip()
df2= pd.DataFrame([df1[questions].corrwith(df1['Pressure_Evaluation'],method='spearman')],index=['Parental Pressure'])
sns.heatmap(df2,cmap='jet',annot=True)
plt.title('Heatmap to show how much do the categories affect Parental Pressure')
plt.show()
```

```python
df1.columns = df1.columns.str.strip()
df2= pd.DataFrame([df1[questions].corrwith(df1['Toxicity_Relevance'],method='spearman')],index=['Parental Toxicity'])
sns.heatmap(df2,cmap='coolwarm',annot=True)
plt.title('Heatmap to show how much the categories affect Parental Toxicity')
plt.show()
```

**6.2 Findings and Final Interpretation**

The correlation analysis revealed different patterns of association for parental pressure and perceived parental toxicity.

For overall parental pressure, Pressure Relevance showed the strongest positive association (ρ = 0.68), followed by Decision Relevance (ρ = 0.63) and Comparison Relevance (ρ = 0.62). Career Relevance (ρ = 0.58), Friendship Relevance (ρ = 0.56), and Personal Decision Relevance (ρ = 0.51) also showed positive associations. Fear Relevance (ρ = 0.36) and Feeling Relevance (ρ = 0.22) showed comparatively weaker associations.

For perceived parental toxicity, Feeling Relevance showed the strongest positive association (ρ = 0.61), followed by Decision Relevance (ρ = 0.56) and Pressure Relevance (ρ = 0.47). Comparison Relevance (ρ = 0.43), Personal Decision Relevance (ρ = 0.42), Career Relevance (ρ = 0.41), and Friendship Relevance (ρ = 0.38) showed positive associations of comparatively moderate magnitude. Guilt Relevance (ρ = 0.19) and Fear Relevance (ρ = 0.18) showed weaker associations.

Final verdict: The results suggest that the experiences associated with overall parental pressure are not identical to those associated with perceived parental toxicity. Pressure Relevance, Decision Relevance, and Comparison Relevance showed the strongest associations with parental pressure, whereas Feeling Relevance and Decision Relevance showed the strongest associations with perceived toxicity. Decision Relevance was among the more strongly associated items for both outcomes.

These findings suggest that parental pressure and perceived parental toxicity may reflect different aspects of respondents' experiences. However, the correlation coefficients alone do not establish causation or statistical significance. The conclusions should therefore be understood as patterns observed among the 142 survey participants, rather than definitive explanations of parental pressure or toxicity.

## **13. Limitations**

Despite providing useful exploratory insights into parental pressure and perceived parental toxicity, this study has several limitations that should be considered when interpreting its findings.

**13.1 Voluntary Convenience Sampling**

The study employed voluntary convenience sampling, in which participation depended on students' willingness to respond to the survey. Collecting responses was challenging because participation was voluntary, limiting the final sample size to ***142 respondents***. Since participants were not selected through random sampling, the collected responses may not fully represent the broader student population. Consequently, the findings should be interpreted with caution when generalising them to other groups.

**13.2 Demographic Imbalance in the Sample**

The sample exhibited demographic imbalances across age, educational level, and gender. Most respondents belonged to the ***20–23 age group***, while the educational distribution was heavily concentrated among undergraduates. Of the 142 respondents, ***128 were undergraduates, 10 were school students, and only 4 were postgraduates***. Additionally, only ***27 respondents were female***, indicating an imbalance in gender representation. These disparities may have limited the study's ability to make reliable comparisons across demographic groups, particularly those represented by very few participants.

**13.3 Self-Reported Ratings**

The study relied on self-reported responses, meaning that the findings reflect participants' personal perceptions and interpretations of parental pressure and toxicity. Such ratings may be influenced by individual experiences, memory, emotional state, personal beliefs, and differences in how respondents understand the survey questions. Therefore, the responses may not always represent objective or independently verified measures of parental behaviour.

**13.4 Observational and Cross-Sectional Design**

The study followed an observational, cross-sectional design, with responses collected at a particular point in time without manipulating any variables. Consequently, the analysis can identify patterns and associations between survey responses and demographic characteristics, but it cannot establish causal relationships. Furthermore, the study does not capture how perceptions of parental pressure or toxicity may change over time.

**13.5 Small Subgroup Sizes**

Although the overall sample consisted of 142 respondents, some demographic subgroups were represented by very few participants, particularly postgraduates, school students, and female respondents. The small sizes of these subgroups limit the reliability of subgroup-level comparisons and may make their results more sensitive to individual responses. Therefore, differences observed between demographic groups should be treated as preliminary rather than conclusive.

**13.6 Potential Measurement Limitations**

The study used survey ratings to represent complex experiences such as parental pressure and perceived parental toxicity. These concepts may involve multiple dimensions that cannot necessarily be captured fully by a limited set of rating-scale questions. Differences in respondents' interpretations of the questions may also introduce measurement variability. Additionally, the limited sample size and the characteristics of the collected data may have affected the performance of the predictive models, with some models performing less effectively than others. The results should therefore be interpreted in light of these measurement and data-related constraints.

**13.7 Exploratory Nature of the Machine Learning Analysis**

The machine learning analysis was conducted as an exploratory investigation rather than as a definitive predictive study. The dataset was divided into training and testing subsets using an 80:20 split, resulting in ***113 observations for training and 29 observations for testing***. These relatively small subsets limit the amount of information available for model training and the reliability of performance evaluation.

With only 29 observations in the test set, the reported performance metrics may be sensitive to the particular train-test split and may not accurately reflect how the models would perform on new, unseen data. Although some models demonstrated more promising results than others, the available evidence is insufficient to establish their general predictive reliability. The machine learning findings should therefore be considered ***preliminary and exploratory***, and further validation using a larger and more representative dataset would be necessary before drawing stronger conclusions.

**13.8 Overall Limitation**

Taken together, these limitations mean that the study is best understood as an initial exploration of students' perceptions of parental pressure and toxicity. The findings provide useful preliminary observations and indicate potential patterns worth investigating further; however, they should not be treated as universally representative or conclusive. Future research could address these limitations by collecting a larger and more demographically balanced sample, refining the measurement instruments, and validating the machine learning models using more extensive data and robust evaluation procedures.

At the same time, the weak correlations with age indicate that the observed differences across educational groups cannot simply be reduced to age-related differences.

Finally, the machine learning analysis demonstrated the potential for predictive modelling while also highlighting the limitations imposed by the size and composition of the dataset.

Above all, these findings should be understood as an exploration of the experiences reported by the 142 participants, not as a definitive description of all students or families. The study identifies patterns that merit further investigation and provides a foundation for future research involving larger, more representative samples and stronger validation methods.

## **15. Conclusion and Future Work**

**15.1 Conclusion88

This study began with an attempt to understand the experiences associated with parental pressure and perceived parental toxicity. However, the findings have raised a broader question: **Are we helping students grow into independent individuals, or are we placing expectations upon them that may prevent them from discovering their own potential?**

In the Indian context, academic marks are often treated as a major indicator of a student's ability and future prospects. The findings of this study suggest that academic expectations, comparisons with other children, and parental involvement in decision-making are strongly associated with the parental-pressure ratings reported by the respondents. Among the experiences examined, pressure to achieve higher examination marks showed the strongest association with parental pressure (\(\rho = 0.68\)), followed by parental decision-making (\(\rho = 0.63\)) and comparison with other children (\(\rho = 0.62\)). Career-related pressure was also associated with the overall pressure rating (\(\rho = 0.58\)).

These results invite us to reconsider the way academic achievement is understood. **Marks are an outcome of learning, not a complete measure of a student's ability, potential, or worth.** Students need encouragement, resources, and guidance to perform well, but excessive pressure to achieve particular scores may undermine the very confidence and independence that education is supposed to develop. Parents can help their children recognise their strengths, understand their weaknesses, and make responsible choices without making every examination result a measure of personal success or failure.

Comparison presents a related concern. Every student has a different background, access to opportunities, interests, abilities, and pace of development. Comparing one child with another may create an expectation that every individual must follow the same path to success. The results of this study suggest that such comparisons are closely associated with reported parental pressure. Instead of measuring a child's progress against the achievements of others, families could place greater emphasis on individual growth, effort, learning, and well-being.

Another important concern is the extent to which young people are allowed to make decisions for themselves. The association between parental decision-making and parental pressure was particularly notable, even though most respondents belonged to the 20–23 age group. This raises an important question about autonomy: **When should parental guidance make room for a young adult's own decisions?** Parents have valuable experience and can help their children understand the consequences of different choices. Nevertheless, guidance should not automatically become control. As children grow older, they need opportunities to explore their interests, evaluate alternatives, accept responsibility, and learn from their decisions. Independence does not require parents to withdraw their support; it requires them to respect their children's growing capacity to decide for themselves.

The findings relating to perceived parental toxicity reveal another dimension of the parent-child relationship. The dismissal of children's feelings during disagreements showed the strongest association with the perceived parental-toxicity rating (\(\rho = 0.61\)). Parental decision-making also showed a notable association (\(\rho = 0.56\)). These results suggest that the emotional experience of being heard and respected deserves as much attention as the expectations placed on students.

Disagreements between parents and children are inevitable. However, disagreement need not mean that one person's feelings must be ignored or that one side must always impose its decision. Listening to a child does not mean agreeing with every choice the child makes. Rather, it means acknowledging the child's feelings, understanding the reasoning behind their perspective, and responding with respect. A family environment in which young people can speak openly without feeling dismissed may provide a healthier foundation for communication and mutual understanding.

**Gender, equality, and the need for equal opportunity**

The gender-wise analysis revealed differences in the distributions of parental-pressure and perceived parental-toxicity ratings. Female respondents' pressure ratings extended further into the higher range, although the median pressure rating was the same for male and female respondents. For perceived toxicity, the female respondents' ratings extended into higher values, but their median was lower than that of male respondents. These differences must be interpreted carefully, particularly because the sample contained substantially more male than female respondents.

Nevertheless, the findings provide a reason to examine whether expectations and experiences within families differ according to gender. In the Indian social context, concerns about gender discrimination, restrictions on personal choices, marriage expectations, and unequal opportunities deserve serious attention. Such experiences were not all directly measured by this survey, so their contribution to the observed ratings cannot be established from the present data alone. They remain important questions for future investigation.

Every child, regardless of gender, deserves respect, educational opportunity, emotional support, and the freedom to develop their capabilities. Discrimination should first be recognised and reduced wherever it occurs, with the longer-term objective of eliminating unequal treatment. Parents and communities should encourage children to pursue their aspirations based on their interests and abilities rather than limiting their choices through gender-based expectations.

**Living background and unequal pressures**

One of the notable observations was that respondents from semi-rural backgrounds reported the highest average parental-pressure and perceived parental-toxicity ratings among the three living-background groups examined. In contrast, the rural group recorded lower average ratings on both measures. These results challenge the assumption that parental pressure necessarily decreases as one moves from rural to urban settings.

The reasons behind this pattern remain uncertain. One possible explanation worth investigating is the relationship between economic circumstances, educational opportunities, family aspirations, and the perceived importance of academic achievement as a route to a more secure future. Families facing financial uncertainty or limited opportunities may place particularly high expectations on education. However, the present survey did not establish that these circumstances caused the observed differences, and the results should not be used to generalise about all rural or semi-rural families.

The findings nevertheless raise a broader social question. If education is viewed as one of the few available routes towards economic security, the expectations placed on children may become especially intense. Addressing this concern requires more than asking parents to reduce pressure. It also requires attention to the conditions in which families make educational and career decisions: access to quality education, awareness of diverse career pathways, financial security, and equal opportunities for students from different backgrounds.

A child's future should not have to carry the entire burden of a family's hopes for a better life. Families deserve support, and students deserve the opportunity to learn and develop without being made solely responsible for overcoming every social or economic constraint.

**Educational stage and individual development**

The educational-level analysis showed that school students recorded the highest average parental-pressure rating, while postgraduate respondents recorded the highest average perceived parental-toxicity rating. Undergraduate respondents reported lower average toxicity ratings than the other two groups. These findings indicate that pressure and perceived toxicity may follow different patterns at different educational stages.

School years are often associated with examinations, academic competition, and preparation for future educational opportunities. It is understandable that parents want their children to have a secure future. However, this concern should not prevent students from discovering their interests, developing practical abilities, or learning from setbacks. Educational guidance should help students make informed choices rather than create the impression that one examination or one career path determines their entire future.

The postgraduate findings must be interpreted with particular caution because only four postgraduate respondents participated in the study. Their responses are insufficient to support broad conclusions about postgraduate students as a group. Similarly, the relatively weak correlations between age and the two outcome measures suggest that age alone did not show a strong association with reported pressure or perceived toxicity in this sample. Further research would be needed to understand how these experiences evolve across different stages of life.

**The role of machine learning**

The machine learning component explored whether the available survey responses could be used to predict overall parental-pressure and perceived parental-toxicity ratings. Linear Regression was selected for parental pressure, while Random Forest Regression was selected for perceived parental toxicity based on the model evaluations conducted.

The parental-pressure Linear Regression model achieved an \(R^2\) score of \(0.44\) and an MAE of \(1.32\) on the test set. These results suggest that the available features contained some predictive information, but they do not establish that the model is sufficiently reliable for real-world decisions. The limited dataset, small test set, and exploratory evaluation restrict the strength of the conclusions that can be drawn.

Machine learning should therefore be viewed as a potential analytical aid rather than a solution to parental pressure or toxicity in itself. A predictive model cannot replace meaningful communication, professional support where needed, or an understanding of each student's individual circumstances. Its value lies in helping researchers investigate patterns, formulate questions, and identify directions for further study—not in labelling families or making definitive judgements about individual relationships.

**Final reflection**

Ultimately, this study is not an attempt to portray parents as the problem or to deny the importance of parental guidance. Parents often act out of concern for their children's future, and academic expectations may arise from a genuine desire to provide security and opportunity. The concern is not whether parents should guide their children, but **how that guidance is expressed and whether it leaves room for the child's feelings, individuality, and independence**.

The findings suggest that academic expectations, comparisons, decision-making autonomy, and emotional validation deserve closer attention in discussions about parental pressure and perceived parental toxicity. They also reveal patterns across demographic groups that merit further investigation.

The study is limited to 142 self-selected respondents and cannot represent the experiences of every student in India. Nevertheless, it offers a starting point for a conversation that deserves greater attention: a conversation about balancing ambition with empathy, guidance with independence, and family expectations with the well-being of young people.

A student's worth cannot be reduced to examination marks, a comparison with another child, or the career that someone else chooses for them. Education should help young people become capable of making their own informed decisions, understanding their responsibilities, and developing their potential. Families can play a vital role in that process by offering support without unnecessary pressure and by treating their children's feelings as worthy of consideration.

The larger objective should not be to eliminate parental guidance, but to encourage a form of parenting in which ambition and affection can coexist, expectations do not overshadow emotional well-being, and children are prepared not merely to meet the expectations of others, but to build thoughtful and independent lives of their own.

**15.2 Proposed Solutions**

Based on the patterns observed in this study, the following recommendations are proposed for consideration. They are suggestions for potential improvement, not interventions whose effectiveness has been demonstrated by this survey.

1. **Encourage learning over marks alone.** Parents and educational institutions should recognise effort, conceptual understanding, improvement, and practical skills alongside examination performance. Academic results should inform guidance rather than become the sole measure of a student's potential.

2. **Reduce harmful comparisons.** Students should be encouraged to measure progress against their own previous performance and personal goals. Comparisons with other children should not be used as a substitute for understanding individual strengths, challenges, and circumstances.

3. **Respect age-appropriate decision-making autonomy.** As children mature, parents should involve them in decisions concerning education, career pathways, and personal development. Discussion and guidance should take precedence over automatically imposing a decision.

4. **Practise emotional validation.** Parents should listen to their children's concerns, acknowledge their feelings during disagreements, and create opportunities for open communication. Listening respectfully does not require parents to agree with every decision.

5. **Promote gender equality and freedom of choice.** Families and educational institutions should discourage gender-based restrictions on education, careers, and personal aspirations. Awareness initiatives should address discrimination and encourage equal opportunities for all students.

6. **Strengthen educational and career guidance.** Schools and colleges can help students and parents understand the range of available academic and professional pathways. Access to reliable information and counselling may help families make decisions based on a student's interests and circumstances rather than fear or social comparison.

7. **Address differences in access to opportunity.** Further attention should be given to educational resources, financial constraints, and career awareness across different living backgrounds. Reducing pressure may require improving the opportunities available to students, not simply asking families to change their expectations.

8. **Encourage further research and informed intervention.** The patterns identified in this study can guide the development of parent-awareness programmes, student-support initiatives, and future research. Any such intervention should be evaluated independently before its effectiveness is claimed.

**15.3 Future Work**

The present study provides a preliminary foundation for further investigation. Several improvements could strengthen future research:

- **Larger and more representative sampling:** Collect responses from a larger number of participants across age groups, educational levels, genders, and rural, semi-rural, and urban backgrounds. Where feasible, use a structured sampling strategy to reduce selection bias.
- **Improved demographic balance:** Recruit sufficient participants from underrepresented groups so that comparisons are not dominated by the largest categories.
- **Refinement of measurement instruments:** Review the survey questions for clarity and coverage of the concepts being studied. Where appropriate, use established instruments and assess their reliability and validity.
- **Deeper statistical analysis:** Examine the uncertainty around observed associations and group differences, using suitable statistical tests, effect sizes, and confidence intervals. Account for multiple comparisons where necessary.
- **Longitudinal research:** Collect responses at multiple points in time to investigate how perceptions of parental pressure and toxicity may change as students progress through education and become more independent.
- **Qualitative investigation:** Conduct interviews or open-ended surveys to understand the experiences behind the numerical ratings, including how students interpret academic expectations, emotional invalidation, comparison, and autonomy.
- **Improved machine learning evaluation:** Use cross-validation, compare models under consistent conditions, conduct hyperparameter tuning where justified, and perform feature engineering only when it is appropriate to the research question. Evaluate both predictive performance and uncertainty using a larger independent test set.
- **Careful examination of predictor-target relationships:** Verify that the input features do not simply reproduce or overlap with the outcome ratings in a way that inflates predictive performance. Any model intended for practical use should be tested on genuinely unseen data.
- **Evaluation of proposed interventions:** Future studies could assess whether parent-awareness initiatives, counselling, career guidance, or improved communication practices are associated with measurable improvements in students' reported experiences.

With these improvements, future research could move beyond identifying preliminary patterns towards developing a more reliable understanding of parental pressure and perceived parental toxicity, the circumstances associated with them, and the approaches that may help support healthier parent-child relationships.

The ultimate aim is not merely to predict ratings, but to understand the experiences behind them and to contribute, through further evidence and thoughtful action, to an environment in which students can pursue their aspirations without feeling that their emotional well-being and independence must be sacrificed to achieve them.
